```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x:x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df['Previous Purchases']=df.groupby(['Category','Frequency of Purchases', 'Subscription Status'])['Previous Purchases'].transform(lambda x:x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df['Purchase Amount (USD)']=df.groupby(['Category','Season','Subscription Status'])['Purchase Amount (USD)'].transform(lambda x:x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df['Size'] = df['Size'].fillna(df['Size'].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
#create a column age_group
labels = ['Young Adult','Adult','Middle-aged','Senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','age_group']].head(10)
```

```python
# create a column purchase_frequency_days
frequency_mapping={
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months':90
}
df['purchase_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
```
